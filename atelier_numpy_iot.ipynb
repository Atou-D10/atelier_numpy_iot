# Partie 1 : Listes Python vs tableaux NumPy
On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8

1) Créer une liste Python stockant ces températures;

```python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperatures_liste)
```

2) Créer un tableau NumPy équivalent stockant ces températures

```python
import numpy as np

temperatures_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_array)
```

3) Effectuer une multiplication par 2 sur ces deux collections

```python
# Sur la liste Python
temperatures_liste_x2 = temperatures_liste * 2
print(temperatures_liste_x2)

print("----------------")

# Sur le tableau NumPy
temperatures_array_x2 = temperatures_array * 2
print(temperatures_array_x2)
```

4) Quelle différence observe-t-on ?

Sur la liste python lorsqu'on effectue une multiplication, les elements de la liste se duplique et quand on fait une multiplication sur le tableau numpy on obtient une multiplication element par element

# Partie 5 – Filtrage booléen
Considérant le tableau représentant les 24 températures d'une journée, on souhaite détecter les
situations inhabituelles.

```python
# 1) Températures supérieures à 30°C
print(temperatures_jour[temperatures_jour > 30])
```

```python
# 2) Températures normales (entre 20°C et 30°C)
print(temperatures_jour[(temperatures_jour >= 20) & (temperatures_jour <= 30)])
```

```python
# 3) Tableau booléen des anomalies (> 35°C)
anomalies = temperatures_jour > 35
print(anomalies)
```

```python
# 4) Nombre de températures anormales
nombre_anomalies = np.sum(anomalies)
print(nombre_anomalies)
```

# Partie 7 – Concaténation

On suppose qu’on a un 2nd bâtiment avec 3 capteurs.

```python
# 1) Matrice des températures du 2nd bâtiment : 3 capteurs x 24 heures
np.random.seed(42)
temperatures_batiment2 = np.random.uniform(20, 30, (3, 24))
print(temperatures_batiment2)
```

```python
# 2) Fusion des 8 capteurs 

# Méthode 1 : np.concatenate avec axis=0
# axis=0 signifie qu'on empile le long des lignes (donc on ajoute des capteurs

temperatures_8_capteurs_v1 = np.concatenate((temperatures_capteurs, temperatures_batiment2), axis=0)
print(temperatures_8_capteurs_v1)
```

# Partie 8 : Calcul scientifique
Considérons le tableau représentant les 24 températures du 1er bâtiment.

1) Pour les températures, calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l'écart-type

```python
# 1) Statistiques descriptives sur temperatures_jour
minimum = np.min(temperatures_jour)
maximum = np.max(temperatures_jour)
somme = np.sum(temperatures_jour)
moyenne = np.mean(temperatures_jour)
mediane = np.median(temperatures_jour)
variance = np.var(temperatures_jour)
ecart_type = np.std(temperatures_jour)

print("Minimum :", minimum)
print("Maximum :", maximum)
print("Somme :", somme)
print("Moyenne :", moyenne)
print("Médiane :", mediane)
print("Variance :", variance)
print("Écart-type :", ecart_type)
```

```python
# 2) Standardisation
temperatures_standardisees = (temperatures_jour - moyenne) / ecart_type
print(temperatures_standardisees)

# Vérification
print("Moyenne après standardisation :", np.mean(temperatures_standardisees))
print("Écart-type après standardisation :", np.std(temperatures_standardisees))
```

```python
# 3) Normalisation Min-Max
temperatures_normalisees = (temperatures_jour - minimum) / (maximum - minimum)
print(temperatures_normalisees)

# Vérification
print("Minimum après normalisation :", np.min(temperatures_normalisees))
print("Maximum après normalisation :", np.max(temperatures_normalisees))
```

```python
# 4) Conversion en Fahrenheit
# Le Fahrenheit est une échelle de température où l'eau gèle à 32°F et bout à 212°F
# (contre 0°C et 100°C en Celsius). La formule de conversion est F = C * 9/5 + 32
temperatures_fahrenheit = temperatures_jour * 9/5 + 32
print(temperatures_fahrenheit)
```
