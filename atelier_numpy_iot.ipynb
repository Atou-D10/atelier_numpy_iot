# Partie 1 : Listes Python vs tableaux NumPy
On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8

1) Créer une liste Python stockant ces températures;

```python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperatures_liste)
```

2) Créer un tableau NumPy équivalent stockant ces températures

```python
import numpy as np

temperatures_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_array)
```

3) Effectuer une multiplication par 2 sur ces deux collections

```python
# Sur la liste Python
temperatures_liste_x2 = temperatures_liste * 2
print(temperatures_liste_x2)

print("----------------")

# Sur le tableau NumPy
temperatures_array_x2 = temperatures_array * 2
print(temperatures_array_x2)
```

4) Quelle différence observe-t-on ?

Sur la liste python lorsqu'on effectue une multiplication, les elements de la liste se duplique et quand on fait une multiplication sur le tableau numpy on obtient une multiplication element par element

Partie 2 – Création des données

1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro.

```python
#  Tableau de 10 capteurs, initialement à zéro
temperatures_capteurs = np.zeros(10)
print(temperatures_capteurs)
```

2) Créer un tableau contenant 10 valeurs égales à 1.

```python
# Tableau de 10 valeurs égales à 1
temperatures_uns = np.ones(10)
print(temperatures_uns)
```

3) Créer un tableau de 10 valeurs initialisées à 25.

```python
# Tableau de 10 valeurs initialisées à 25
temperatures_25 = np.full(10, 25)
print(temperatures_25)
```

4) Créer une matrice identité de taille 5 × 5.

```python
#  Matrice identité 5x5
matrice_identite = np.eye(5)
print(matrice_identite)
```

5) Créer les 24 heures (de 0 à 23) de mesure d’une journée

```python
# Les 24 heures de la journée (0 à 23)
heures = np.arange(24)
print(heures)
```

6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C

```python
# 10 valeurs régulièrement réparties entre 20°C et 30°C
temperatures_reparties = np.linspace(20, 30, 10)
print(temperatures_reparties)
```

7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles

```python
# 100 mesures aléatoires de température, reproductibles
np.random.seed(42)
temperatures_100 = np.random.uniform(20, 30, 100)
print(temperatures_100)
```

# Partie 3 – Exploration des tableaux

1) Créer aléatoirement un tableau représentant les 24 températures d'une journée

```python
# 24 températures d'une journée (aléatoire)
np.random.seed(42)
temperatures_jour = np.random.uniform(20, 30, 24)
print(temperatures_jour)
```

2) Déterminer les propriétés suivantes du tableau
    a) nombre de dimensions 
    b) nombre de valeurs 
    c) type des données 
    d) forme du tableau

```python
# 2) Propriétés du tableau
print("a) Nombre de dimensions :", temperatures_jour.ndim)
print("b) Nombre de valeurs :", temperatures_jour.size)
print("c) Type des données :", temperatures_jour.dtype)
print("d) Forme du tableau :", temperatures_jour.shape)
```

# Partie 4 – Indexation et slicing

```python
# 2) La dernière mesure
print(temperatures_jour[-1])
```
